# Theorie

Volgens [Newton's wet van afkoeling](https://nl.wikipedia.org/wiki/Afkoelingswet_van_Newton) is de snelheid waarmee een voorwerp afkoelt evenredig met het verschil in de temperatuur van het voorwerp ($T$) en de omgeving ($T_0$). We kunnen dit schrijven als:

$$
    \dot{Q} = -hA(T(t) - T_0),
$$

waarin 

- $\dot{Q}$ de warmtestroom in $\mathrm{W}$,
- $A$ het oppervlak waardoor koeling optreedt in $\mathrm{m}^2$,
- $h$ de warmteoverdrachtscoëfficiënt in $\mathrm{W/(m^2 K)}$.

Dit levert de differentiaalvergelijking

$$
    C\dot{T} = -hA(T(t) - T_0),
$$

met $C$ de warmtecapaciteit in $\mathrm{J/kg}$. Herschrijven met $\tau = \frac{C}{hA}$ levert:

$$
    -\tau\dot{T} = T(t) - T_0,
$$ (eq:dv1)

met als oplossing:

$$
    T(t) - T_0 = (T(0) - T_0)\text{e}^{-t/\tau}.
$$ (eq:dv1_opl)

We kunnen hieruit dus concluderen dat $\tau$ de karakteristieke tijdsduur is waarin de temperatuur van de buis een factor $\text{e}$ verlaagd ten opzichte van de omgevingstemperatuur. 

```{exercise} Oplossing DV controleren 
:label: ex-dvopl
Controleer bovenstaande door vergelijking [](#eq:dv1_opl) te substitueren in [](#eq:dv1).
```

````{solution} ex-dvopl
Neem de oplossing van Newtons afkoelingswet in de vorm T(t)=T_0+(T(0)-T_0)e^(-t/tau).  Differentieer deze uitdrukking naar t: omdat T_0 constant is, geldt T_punt(t)=d/dt[(T(0)-T_0)e^(-t/tau)]=(T(0)-T_0)(-1/tau)e^(-t/tau). Vermenigvuldig nu met -tau: -tau T_punt(t)=-tau(-(T(0)-T_0)(1/tau)e^(-t/tau))=(T(0)-T_0)e^(-t/tau). Maar uit de gegeven oplossing volgt ook direct dat T(t)-T_0=(T(0)-T_0)e^(-t/tau). Dus is (-tau T_punt(t)=T(t)-T_0), en daarmee voldoet de oplossing aan de differentiaalvergelijking.
````

We zijn hier voor het gemak uitgegaan van een $h$ die onafhankelijk is van de temperatuur. We weten echter dat warmtetransport door straling niet lineair gaat, maar als 

$$
    \dot{Q}_s = \epsilon \sigma A (T^4 - T_0^4).
$$ 

Voor kleine temperatuurverschillen ($\Delta T = T - T_0$) is dit te vereenvoudigen tot

$$
    \dot{Q}_s = \epsilon \sigma A ((T_0+\Delta T)^4 - T_0^4) \approx \epsilon \sigma 4A T_0^3 \Delta T.
$$

Zolang $\Delta T$ dus relatief klein is ten opzichte van $T_0$, kunnen we $h$ dus inderdaad als een constante beschouwen. 

```{exercise} Uitwerken
:label: ex_uitw

Laat zien dat bovenstaande geldt door eerst $(T_0+\Delta T)^4$ uit te schrijven, te substitueren en dan te bedenken dat $\Delta T$ klein is en $\Delta T^2$ dus nog kleiner.
```

```{solution} ex_uitw
antwoord:

De stralingswarmtestroom wordt gegeven door de Stefan–Boltzmann-relatie Q_punt_s=epsilon*sigma*A(T^4-T_0^4). Neem T=T_0+Delta_T en werk de vierde macht uit:(T_0+Delta_T)^4=T_0^4+4T_0^3*Delta_T+6T_0^2*Delta_T^2+4T_0*Delta_T^3+Delta_T^4. Als je dit invult in (T_0+Delta_T)^4-T_0^4, vallen de T_0^4-termen weg en blijft over:(T_0+Delta_T)^4-T_0^4=4T_0^3*Delta_T+6T_0^2*Delta_T^2+4T_0*Delta_T^3+Delta_T^4. Omdat Delta_T klein is ten opzichte van T_0 zijn Delta_T^2, Delta_T^3, Delta_T^4 veel kleiner dan Delta_T, en mag je die hogere-ordetermen verwaarlozen; dan volgt de linearisatie Q_punt_s ongeveer gelijk aan 4*epsilon*sigma*A*T_0^3*Delta_T.
```


```{exercise} 
:label: ex_fout
Hoe groot is de fout in het warmtetransport door straling die we maken voor het temperatuurbereik waarin we deze proef uitvoeren?
```

```{solution} ex_fout
antwoord:

De fout die wordt gemaakt in het warmtetransport door straling wordt geschat door de verwaarloosde termen in de Taylor-reeks te bekijken. Voor kleine Delta_T ten opzichte van T_0, is de grootste verwaarloosde term 6*T_0^2*Delta_T^2. De relatieve fout in het warmtetransport door straling is dan ongeveer (6*T_0^2*Delta_T^2) / (4*T_0^3*Delta_T) = (3/2)*(Delta_T/T_0). Stel dat T_0 ongeveer 300 K is en Delta_T maximaal 10 K, dan is de relatieve fout ongeveer (3/2)*(10/300) = 0.05, oftewel 5%. Dus de fout in het warmtetransport door straling die we maken voor het temperatuurbereik waarin we deze proef uitvoeren is ongeveer 5%.
```

# Resultaten

`````python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt

times = np.array([0,10,20,30,40,50,60,70,90,100,110,120,130,140,150,160,170,180,190,200,210,220,230,240,250,260,270], dtype=float)
temps = np.array([49.7,49.5,48.7,48.2,47.4,46.6,46.0,45.3,44.0,43.4,42.7,42.3,41.7,41.3,40.8,40.4,40.0,39.6,39.1,38.7,38.4,38.1,37.7,37.4,37.0,36.7,36.4], dtype=float)

T0 = 21.5  # temperatuur omgeving

def exp_fixed_T0(t, A, tau):
    return A*np.exp(-t/tau) + T0

# fit met vaste omgevingstemperatuur
popt, pcov = curve_fit(exp_fixed_T0, times, temps, p0=[temps[0]-T0, 300], maxfev=10000)
A_exp, tau_exp = popt

t_plot = np.linspace(times.min(), times.max(), 300)
T_fit = exp_fixed_T0(t_plot, A_exp, tau_exp)

plt.figure()
plt.plot(times, temps, "bo", label="metingen")
plt.plot(t_plot, T_fit, "r-", label="fit")
plt.xlabel("tijd [s]")
plt.ylabel("temperatuur [°C]")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# afmetingen koelbuis
L = 0.1531 # lengte in m
D = 0.05 # diameter in m 
r = D/2 # buitensstraal in m 
t = 0.0015 # wanddikte in m
ri = r - t # binnensstraal in m

# oppervlakte berekenen
A_mantel = np.pi*D*L
A_einden = 2*np.pi*r**2
A = A_mantel + A_einden  

# materiaal (messing)
rho = 8.73e3 # in kg/m^3
c = 3.8e2 # in J/(kg*K)

# massa en warmtecapaciteit berekenen
V = np.pi*(r**2 - ri**2)*L # in m^3
m = rho*V # in kg
C = m*c  # in J/K

h = C/(tau_exp*A) # in W/(m^2*K)

# onzekerheden bepalen
A_u, tau_u = np.sqrt(np.diag(pcov))
h_u = h * (tau_u/tau_exp)

print("De warmteoverdrachtscoefficient is", h, "+/-", h_u, "W/(m^2 K)")
`````

# Conclusie

De warmtecapaciteit van het koelbuis systeem is bepaald door de massa en de specifieke warmtecapaciteit van het materiaal (messing). De berekende warmteoverdrachtscoefficient is 10.281793968429778 +/- 0.11482384933133961 W/(m^2 K)

# Discussie

De meeste fouten komen niet door de fit van de data, maar door andere factoren zoals de schatting van de warmtecapaciteit, onnauwkeurigheden in de temperatuurmeting en de schatting van de omgevingstemperatuur. Daarnaast kan de aanname dat de warmteoverdrachtscoëfficiënt constant is, leiden tot afwijkingen, vooral als de temperatuurverschillen groter zijn dan verwacht. Het is ook mogelijk dat er extra warmteverliezen zijn door straling of convectie van de lucht die niet zijn meegenomen in de berekeningen. Toekomstige experimenten zouden meetapparatuur die accurater zijn kunnen gebruiken en een beter gecontroleerde omgeving om deze fouten te verminderen.
