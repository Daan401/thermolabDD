# Is het verdamping?

In dit experiment is er een verwarmingselement in een met water gevulde maatbeker gestopt. Elke minuut is de temperatuur van het water gemeten. Deze metingen zijn opgeslagen in [tempmetingen.csv](tempmetingen.csv).

Verder is gegeven dat: 
$m_{maatbeker} = 820.8 \mathrm{gr}$  
$m_{maatbeker + water, start} = 1292.9 \mathrm{gr}$  
$m_{maatbeker + water, eind} = 1274.9 \mathrm{gr}$

Het moge duidelijk zijn dat er water is 'verdwenen'. Ook is de eindtemperatuur van het water niet gelijk aan een proces waarbij verdamping en eventuele warmteverliezen niet meegenomen worden.

```{exercise}
Onderzoek het proces door:
1. De plot van de metingen te maken, inclusief een verwachte trendlijn voor de temperatuur van het water als al de toegevoerde warmte gaat zitten in het water.
2. Bepaal of de totale hoeveelheid toegevoegde energie correspondeert met de opwarming en verdamping van het water, geef daarbij de aannames die je doet.
3. Geef aanbevelingen voor een verbetering van het experiment om (nog) beter het thermodynamische proces te kunnen onderzoeken en beschrijven.
```

````python
import numpy as np
import matplotlib.pyplot as plt

data = np.genfromtxt(
    "tempmetingen-1f4021492f1702d6ed60d67d81976e9e.csv",
    delimiter=";",
    skip_header=1)

t = data[:, 0]      # tijd in minuten
T = data[:, 1]      # temperatuur in °C


mask = t <= 10
a, b = np.polyfit(t[mask], T[mask], 1)

T_ideaal = a * t + b

plt.figure
plt.plot(t, T, ".", label="Gemeten temperatuur")
plt.plot(t, T_ideaal, "--", label="Ideaal")
plt.xlabel("Tijd (min)")
plt.ylabel("Temperatuur (°C)")
plt.title("Opwarming water met verdamping")
plt.legend()
plt.grid(True)
plt.show()

#benodigde energie voor het verwarmen en verdampen van het water, waarbij wordt uitgegaan van ideale opwarming zonder warmteverlies
T_start = T[0]
T_end = T[-1]
deltaT = T_end - T_start

m_water = 472.1        # gram
c_water = 4.18         # J/(g K)
verdampingswarmte_w = 2256  # J/g
m_damp = 18.0          # gram

Q_opwarming = m_water * c_water * deltaT
Q_verdamping = m_damp * verdampingswarmte_w
Q_totaal = Q_opwarming + Q_verdamping

print(f"Totaal benodigde energie: {Q_totaal:,.0f} J")

#deze gekregen waarde zou je kunnen checken als je weet hoeveel vermogen de warmtebron levert en hoe lang die aan heeft gestaan, want Q = P * t
#verder zie je dat de gemeten temperatuur lager ligt dan de ideale temperatuur, wat wijst op warmteverlies aan de omgeving en dus een niet ideaal proces
````

Verbeteringen voor de volgende keer:
- je zou de maatbeker kunnen isoleren om warmteverlies door straling te verminderen
- Je zou het water tijdens het opwarmen kunnen roeren voor een homogene temperatuurverdeling
- je zou het vermogen van het opwarmingselement kunnen achterhalen, om zo te kijken hoeveel energie er precies nuttig wordt gebruikt voor de opwarming en verdamping van het water
